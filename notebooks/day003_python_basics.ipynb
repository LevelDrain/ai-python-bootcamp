## 📖 講義：AIにとって「データを選ぶ」とは

こんなデータがあるとします。

```python
import numpy as np

scores = np.array([42, 78, 91, 55, 83])
```

3番目の値が欲しいなら、

```python
scores[2]
```

結果は、

```text
91
```

Pythonと同じく**0始まり**ですね。

そしてNumPyでは、もっと強力なことができます。

```python
scores[scores >= 80]
```

結果：

```text
[91 83]
```

この一行では、

```text
scores >= 80
      ↓
[False, False, True, False, True]
      ↓
Trueの場所だけ取得
      ↓
[91, 83]
```

という処理が行われています。

これが **Boolean Mask（真偽値マスク）**。

後でscikit-learnを使うと、

```text
データ
 ↓
異常値を除く
 ↓
特定クラスを抽出
 ↓
訓練データとテストデータに分割
 ↓
モデルへ投入
```

といった処理を大量に行います。

つまり今日は**モデルに食わせるデータを料理する技術**です。🍳

# 💻 実習2：スライス

```python
print(a[1:4])
```

これは、

```text
index    0   1   2   3   4
value   10  20  30  40  50
             └───────┘
```

なので、

```text
[20 30 40]
```

になります。

基本形は、

```python
a[start:stop:step]
```

です。

例えば、

```python
a[::2]
```

なら1個飛ばし。

```text
[10 30 50]
```

## 列を丸ごと取り出す

```python
print(X[:, 1])
```

結果：

```text
[20 50 80 110]
```

この `:` が重要。

```text
X[:, 1]
│ │  └─ 1列目
│ └──── 全行
└────── X
```

つまり**全サンプルについて、特定の特徴量だけ取得**できます。

これは今後めちゃくちゃ使います。

# 💻 実習4：Boolean Mask

本日の主役です。

```python
scores = np.array([42, 78, 91, 55, 83])

mask = scores >= 80

print(mask)
```

結果：

```text
[False False  True False  True]
```

そして、

```python
print(scores[mask])
```

結果：

```text
[91 83]
```

一行でも書けます。

```python
scores[scores >= 80]
```

# 💻 実習5：複数条件

「60以上90未満」を取り出したい。

NumPyではこうします。

```python
scores[(scores >= 60) & (scores < 90)]
```

結果：

```text
[78 83]
```

ここで注意。

Pythonの、

```python
and
```

ではなく、

```python
&
```

を使います。

ORなら、

```python
|
```

です。

そして各条件を **`()` で囲む**。

これはNumPy初心者が一度は踏む地雷です💣

```python
(scores >= 60) & (scores < 90)
```

の形ごと覚えてしまってOK。

# 💻 実習6：実際のデータっぽくする

5人について、

```text
[勉強時間, テスト点数]
```

があるとします。

```python
data = np.array([
    [1.0, 45],
    [2.5, 62],
    [3.0, 70],
    [4.5, 88],
    [6.0, 95]
])
```

点数だけ取得。

```python
scores = data[:, 1]

print(scores)
```

```text
[45. 62. 70. 88. 95.]
```

80点以上の人を抽出します。

```python
high_score = data[data[:, 1] >= 80]

print(high_score)
```

結果：

```text
[[ 4.5 88. ]
 [ 6.  95. ]]
```

おお。

**「点数が80以上の行を丸ごと取得」**

できました。

これ、もう小さいながら**データサイエンスのコード**です。
