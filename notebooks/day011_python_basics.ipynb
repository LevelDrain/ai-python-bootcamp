という役割分担になります。

# 💻 実習2：train_test_split

# 💻 実習3：StandardScaler

まず、

```python
from sklearn.preprocessing import StandardScaler
```

読み込み。

```python
scaler = StandardScaler()
```

ここで、

```python
scaler.fit(X_train)
```

を実行します。

---

## fitとは？

ここ重要。

fitは

```
学習する
```

ではありません。

ここでは、

```text
平均

標準偏差
```

を計算して覚えているだけです。

確認。

```python
print(scaler.mean_)
```

平均が出ます。

さらに、

```python
print(scaler.scale_)
```

標準偏差が確認できます。

つまり、

```python
fit()

↓

平均を記憶

↓

標準偏差を記憶
```

です。

---

# 💻 実習4：transform

次。

```python
X_train_scaled = scaler.transform(X_train)
```

確認。

```python
print(X_train_scaled)
```

さらに、

```python
print(
    X_train_scaled.mean(axis=0)
)
```

およそ

```text
0
```

になります。

---

## fit_transform

毎回

```python
fit()

transform()
```

と書くのは面倒なので、

```python
scaler.fit_transform(X_train)
```

という短縮版があります。

意味は

```
fit

↓

transform
```

です。

---

# 📖 なぜテストデータでfitしない？

これは超重要です。

正しい流れ。

```text
train

↓

fit

↓

平均・標準偏差決定

↓

train transform

↓

test transform
```

です。

つまり、

```python
scaler.fit(X_train)

X_train = scaler.transform(X_train)

X_test = scaler.transform(X_test)
```

となります。

---

もし、

```python
fit(X_test)
```

すると、

テストデータの情報を学習時に見てしまいます。

これを

> **データリーク（Data Leakage）**

と言います。

AI開発では重大なミスです。

---

# 💻 実習5：fitとtransformの違いを見る

```python
print(scaler.mean_)
```

次に、

```python
X_test_scaled = scaler.transform(
    X_test
)

print(X_test_scaled)
```

ここで使われる平均は、

**trainで計算した平均**です。

これが重要。

# 🌿 今日のまとめ

今日はNumPyで自分たちが書いてきた前処理が、scikit-learnではどのようにライブラリ化されているかを学びました。
