# 講義（20分）

## AIは何を見ているの？

例えば猫の画像。

人間には「猫」に見えますが、コンピュータには

154 121  87 ...

> **「数値データをどう加工し、どう計算するか」**

が仕事の中心になります。

---

## Pythonのリストとの違い

普通のPythonでは

```python
numbers = [1, 2, 3]
```

と書けます。

NumPyでは

```python
import numpy as np

numbers = np.array([1, 2, 3])
```

となります。

見た目は似ていますが、

NumPy配列は

* 高速
* メモリ効率が良い
* 数学計算が得意

という特徴があります。

---

## ndarray

NumPyで最も重要なのが

```python
ndarray
```

です。

名前は

```
n-dimensional array
```

つまり

> **多次元配列**

という意味です。

AIではこの型ばかり扱います。

# 次回予告

**第2回：ベクトル演算とブロードキャスト**

NumPyが「ただの配列ライブラリ」ではない理由が見えてきます。ベクトル化やブロードキャストを理解すると、AIコードがシンプルで高速になる理由を体感できるようになります。

````python
import numpy as np

a = np.array([[1, 2, 3]])
print(a)

A = np.array([[1, 2, 3], [4, 5, 6]])
print(A)

print(A.shape)
````

````python
# 問1
# 次の配列を作ろう。
# 1 2 3
# 4 5 6

A = np.array([[1, 2, 3]])
B= np.array([[4, 5, 6]])

# 問2
# shapeを表示してみよう。

print(A.shape)

# 問3
# 3×3の0行列を作ろう。

zero = np.zeros((3, 3))
print(zero)

# 問4
# 5×5の乱数行列を作ろう。

rand = np.random.rand(5, 5)
print(rand)

# 問5
# 次を計算してみよう。
# A+B

print(A + B)

# A*B

print(A * B)

# A@B

print(A @ B.T)

# 違いを確認してみよう。
````
