# 📖 講義：なぜ乱数が必要なの？

AIでは意外なほど乱数を使います。

例えば、

* データをシャッフルする
* 学習データとテストデータに分ける
* ニューラルネットワークの重みを初期化する
* ランダムサンプリングする

などです。

しかし困ることがあります。

今日実験した結果が

```

```

だったのに、

明日同じコードを動かすと

```

```

になる。

それでは研究も開発も困ります。

そこで重要なのが

> **再現性（Reproducibility）**

です。

# 🌿 今日のまとめ

今日は、機械学習実験の基本となる「乱数との付き合い方」を学びました。

重要なのは次の4点です。

* `np.random.seed()`：実験を再現できるようにする。
* `np.random.permutation()`：データを安全にシャッフルする。
* `np.random.choice()`：ランダムサンプリングを行う。
* **インデックスを使って `X` と `y` を一緒に並べ替える**：特徴量と正解ラベルの対応を保つ。

これらは、後で scikit-learn の `train_test_split()` を使うときに、「ライブラリが内部で何をしているのか」を理解する土台になります。

## 次回予告

**第9回：NumPyで探索的データ分析（EDA）の基礎**

いよいよ「データを眺めて理解する」工程に入ります。最小値・最大値・分位数・ヒストグラムの考え方を学び、機械学習を始める前に「まずデータを見る」というAIエンジニア・データサイエンティスト共通の習慣を身につけます。

````python
import numpy as np

# 演習

# 今日のデータ。
Z = np.arange(30)
print(Z)

# 問1
# 乱数seedを42に設定してください。
np.random.seed(42)

# 問2
# `Z` をシャッフルしてください。
idx = np.random.permutation(Z)
print("idx:", idx)
Z=Z[idx]
print(Z)

# 問3
# 7:3に分割してください。
split = int(len(Z)*0.7)
train = Z[:split]
test = Z[split:]
print("split:", split)
print("train:", train)
print("test:", test)

# 問4
# 100人の中から10人を重複なしで抽出してください。
people = np.arange(100)
sample = np.random.choice(
    people,
    size=10,
    replace=False
)
print(sample)

# 問5（ボス戦👾）
# 次があります。

X = np.array([
    [1,10],
    [2,20],
    [3,30],
    [4,40],
    [5,50]
])

y = np.array([
    0,
    1,
    0,
    1,
    0
])

# `X` と `y` の対応を崩さず、
# ランダムに並べ替えてください。

idx2 = np.random.permutation(len(X))
X = X[idx2]
y = y[idx2]
print("idx:", idx2, "/X:", X, "/y:", y)

# | 関数                              | 何をする？            | 戻り値      |
# | ------------------------------- | ---------------- | -------- |
# | `np.random.shuffle(Z)`          | Z自身をシャッフル        | `None`   |
# | `np.random.permutation(Z)`      | シャッフルされた新しい配列を返す | 配列       |
# | `np.random.permutation(len(X))` | シャッフル用インデックスを作る  | インデックス配列 |
````
