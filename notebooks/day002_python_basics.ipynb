# 講義（20分）

## Pythonらしい書き方

例えば全員に1点加点したいとします。

普通のPythonなら

```python
scores = [50, 60, 70]

new_scores = []

for s in scores:
    new_scores.append(s + 1)
```

AIでもできますが、遅いです。

---

NumPyなら

```python
import numpy as np

scores = np.array([50, 60, 70])

print(scores + 1)
```

結果

```
[51 61 71]
```

for文がありません。

NumPyは内部で高速なC言語実装を使っているため、このような書き方が速くなります。

---

## ベクトル化（Vectorization）

この

```
配列同士をまとめて計算する
```

考え方を

**ベクトル化**

と言います。

AIコードでは毎日のように使います。

---

## ブロードキャスト

NumPyには便利な仕組みがあります。

```python
A = np.array([
    [1,2],
    [3,4]
])

print(A + 10)
```

結果

```
11 12
13 14
```

10が自動的に全要素へ広がっています。

これが

**ブロードキャスト**

です。

---

## なぜ重要？

例えば画像。

```
高さ × 幅 × RGB
```

全部の画素に

```
255で割る
```

ことがよくあります。

```python
image = image / 255
```

これだけで終わります。

for文は不要です。

# 今日のポイント

今日はAIコードを書く上で最も重要な考え方を学びました。

* **ベクトル化**：for文ではなく、配列全体を一度に計算する。
