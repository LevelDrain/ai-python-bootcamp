# 💻 実習3：外れ値候補を探す

データを変更します。

```python
scores = np.array([
    68,72,75,80,81,
    85,88,90,92,200
])
```

まず四分位数。

```python
Q1 = np.percentile(scores,25)

Q3 = np.percentile(scores,75)

IQR = Q3 - Q1
```

外れ値候補。

```python
lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR
```

抽出。

```python
mask = (
    (scores >= lower)
    &
    (scores <= upper)
)

print(mask)

print(scores[~mask])
```

ここでも

第3回のBoolean Maskが活躍します。

## 次回予告

**第10回：NumPy総合演習① 〜 前処理パイプラインを作る 〜**

これまで学んだ内容を組み合わせ、

* データ読み込み（コード内で用意）
* 欠損値処理
* 外れ値確認
* 標準化
* シャッフル
* 学習用・評価用データへの分割

までを一つの流れとして実装します。

この回で、NumPy編の前半〜中盤の内容を「点」ではなく「線」として使えるようにしていきます。

````python
import numpy as np

# 今日のデータ。
X = np.array([
    [23,160,55],
    [25,162,58],
    [28,168,60],
    [30,170,65],
    [35,175,72],
    [40,178,80],
    [45,180,85],
    [50,182,90]
], dtype=float)
# 列は年齢、身長、体重です。

# ### 問1
# 列ごとの平均、標準偏差を求めてください。
print("mean:", X.mean(axis=0))
print("std:", X.std(axis=0))

# ### 問2
# 列ごとの最小、最大を求めてください。
print("min:", X.min(axis=0))
print("max:", X.max(axis=0))

# ### 問3
# 身長列だけ取り出してください。
height = X[:, 1]
print("height:", height)

# ### 問4
# 身長について
# Q2：中央値
# Q3：IQR
# を求めてください。
Q2 = np.percentile(height, 50)
Q3 = np.percentile(height, 75)
print("Q2:", Q2)
print("Q3:", Q3)

# ### 問5（ボス戦👾）
# 体重列について
# IQR法で外れ値候補を検出してください。
weight = X[:,2]
wQ1 = np.percentile(weight, 25)
wQ3 = np.percentile(weight, 75)
wIQR = wQ3 - wQ1
lower = wQ1 - 1.5 * wIQR
upper = wQ3 + 1.5 * wIQR
mask = (
    (weight >= lower)
    &
    (weight <= upper)
)
print("mask:", mask)
print("外れ値候補:", weight[~mask])
````
